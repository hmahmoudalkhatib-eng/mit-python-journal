```python
starting_salary = float(input("Enter the starting salary: "))

semi_annual_raise = 0.07
investment_return = 0.04
total_months = 36
down_payment = 250000

low = 0
high = 10000
steps = 0

# Check if it is even possible at 100% savings rate
current_savings = 0
curr_salary = starting_salary
for month in range(1, total_months + 1):
    current_savings += current_savings * (investment_return / 12)
    current_savings += (curr_salary / 12) * 1.0
    if month % 6 == 0:
        curr_salary += curr_salary * semi_annual_raise

if current_savings < down_payment:
    print("It is not possible to pay the down payment in three years.")
else:
    while high - low > 1:
        guess = (high + low) // 2
        rate = guess / 10000.0

        current_savings = 0
        curr_salary = starting_salary

        for month in range(1, total_months + 1):
            current_savings += current_savings * (investment_return / 12)
            current_savings += (curr_salary / 12) * rate
            if month % 6 == 0:
                curr_salary += curr_salary * semi_annual_raise

        if current_savings < down_payment:
            low = guess
        else:
            high = guess

        steps += 1

    print("Best savings rate:", high / 10000.0)
    print("Steps in bisection search:", steps)
```
