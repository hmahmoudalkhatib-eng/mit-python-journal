```python
# Recall the approximation method code to find the... [line cut off]
x = 1121212

epsilon = 1
num_guesses = 0
guess = 0.0
increment = 1    # try it with 0.00001
while abs(guess**2 - x) >= epsilon and guess**2 <= x:  # [rest of line cut off]
    # abs(guess**2 - x) >= epsilon finds a "good... [cut off]
    # guess**2 <= x ensures we stop looking when... [cut off]
    guess += increment
    num_guesses += 1
print(f'num_guesses = {num_guesses}')

# this "if" is for the case when we stopped the loo... [cut off]
if abs(guess**2 - x) >= epsilon:
    print(f'Failed on square root of {x}')
    print(f'Last guess was {guess}')
    print(f'Last guess squared is {guess*guess}')
# this "else" is for the case when we stopped the l... [cut off]
else:
    print(f'{guess} is close to square root of {x}')
```
