```python
ask = int(input("what is your annual salary :"))
ask1 = int(input("what is your portion saved :"))
ask2 = int(input("what is the total cost of your dream home :"))
current_savings = 0
down_payment = 0
r=0.04




months_passed = 0

while current_savings < ask2:
  formula=current_savings*r/12

  down_payment=formula +ask1
  current_savings+=down_payment
  months_passed+=1
print  (months_passed)
```
