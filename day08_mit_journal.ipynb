```python
x=46
epsilon=0.001
guess=0.0
num_guess=0
increment=0.0001
while abs(guess**2 - x)<=epsilon:
  guess+=increment
  num_guess+=1
  print("num_guess =", num_guess)

  print(guess , "is close the the square root of " ,x)
```
